```python
# Converter data
df['data'] = pd.to_datetime(df['data'], format='%d/%m/%Y', errors='coerce')

# verificando a conversão 
print(f"Datas nulas após conversão: {df['data'].isna().sum()}")
```

```python
# Meu interesse é em analisar apenas o campeonato brasileiro de 2025, então vou filtrar apenas este ano
df_2025 = df[df['data'].dt.year == 2025].copy()

print(f"Total de partidas em 2025: {len(df_2025)}")
print(f"Período: {df_2025['data'].min()} até {df_2025['data'].max()}")
```

```python
# Verificar as colunas para compreender melhor o dataset 
df_2025.columns.tolist()
```

```python
# Varificar e tratar valores nulos
nulos = (df_2025.isnull().sum() / len(df_2025) * 100).round(2)
nulos[nulos > 0].sort_values(ascending=False)
```

```python
df_2025.head(20)
```

```python
df_2025 = df_2025.rename(columns={
    'mandante_Placar': 'gols_mandante',
    'visitante_Placar': 'gols_visitante',
    'rodata': 'rodada'
})
```

```python
df_2025.head()
```

```python

df_2025['total_gols'] = df_2025['gols_mandante'] + df_2025['gols_visitante']
```

```python
df_2025['saldo_gols'] = df_2025['gols_mandante'] - df_2025['gols_visitante']
```

```python
def classificar_resultado(row):
    if row['gols_mandante'] > row['gols_visitante']:
        return 'Vitória do mandante'
    elif row['gols_mandante'] < row['gols_visitante']:
        return 'Vitória do visitante'
    else:
        return 'Empate'

df_2025['resultado'] = df_2025.apply(classificar_resultado, axis=1)
```

```python
df_2025['mes'] = df_2025['data'].dt.month
```

```python
print(f'Linhas duplicadas: {df_2025.duplicated(subset=['data', 'mandante', 'visitante']).sum()}')
```
