```python
# Converter data
partidas['data'] = pd.to_datetime(partidas['data'], format='%d/%m/%Y', errors='coerce')

# verificando a conversão 
print(f"Datas nulas após conversão: {partidas['data'].isna().sum()}")
```

```python
# Meu interesse é em analisar apenas o campeonato brasileiro de 2025, então vou filtrar apenas este ano
df_2025 = partidas[partidas['data'].dt.year == 2025].copy()

print(f"Total de partidas em 2025: {len(df_2025)}")
print(f"Período: {df_2025['data'].min()} até {df_2025['data'].max()}")
```

```python
# Verificar as colunas para compreender melhor o dataset 
df_2025.columns.tolist()
```

```python
# Varificar e tratar valores nulos
nulos = (df_2025.isnull().sum() / len(df_2025) * 100).round(2)
nulos[nulos > 0].sort_values(ascending=False)
```

```python
df_2025 = df_2025.rename(columns={
    'mandante_Placar': 'gols_mandante',
    'visitante_Placar': 'gols_visitante',
    'rodata': 'rodada'
})
```

```python
df_2025.head()
```

```python
# Total de gols na partida
df_2025['total_gols'] = df_2025['gols_mandante'] + df_2025['gols_visitante']
```

```python
# Saldo de gols (mandante)
df_2025['saldo_gols'] = df_2025['gols_mandante'] - df_2025['gols_visitante']
```

```python
df_2025['mes'] = df_2025['data'].dt.month
```

```python
# Resultado da partida 
def classificar(gols_a, gols_b):
    if gols_a > gols_b:
        return 'Vitória'
    elif gols_a < gols_b:
        return 'Derrota'
    return 'Empate'

df_2025['resultado_mandante'] = df_2025.apply(
    lambda row: classificar(row['gols_mandante'], row['gols_visitante']), axis=1
)

df_2025['resultado_visitante'] = df_2025.apply(
    lambda row: classificar(row['gols_visitante'], row['gols_mandante']), axis=1
)
```

```python
df_mandante = df_2025[[
    'data', 'rodada', 'mandante', 'tecnico_mandante',
    'gols_mandante', 'gols_visitante', 'resultado_mandante'
]].copy()

df_mandante.columns = ['data', 'rodada', 'time', 'tecnico', 'gols_pro', 'gols_contra', 'resultado']
df_mandante['mando'] = 'Casa'

df_mandante.head(20)
```

```python
df_visitante = df_2025[[
    'data', 'rodada', 'visitante', 'tecnico_visitante',
    'gols_visitante', 'gols_mandante', 'resultado_visitante'
]].copy()

df_visitante.columns = ['data', 'rodada', 'time', 'tecnico', 'gols_pro', 'gols_contra', 'resultado']
df_visitante['mando'] = 'Fora'

df_visitante.head(20)
```
